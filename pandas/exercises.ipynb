# Pandas in 30 short tasks

Each task needs one or two lines of code. Run the setup cell first — it builds the data:

- `df` — results of 205 students in an AI olympiad: city, school, grade, points for five tasks
  (0–100, some missing), Python experience and submission time. A few rows are duplicated on purpose.
- `schools` — school type for every school.

Useful: [pandas cheat sheet](https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf) · `df.columns` · `help(pd.DataFrame.groupby)`

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
n = 200
schools_list = ["III LO", "V LO", "XIV LO", "Technikum nr 1", "Technikum Łączności"]
df = pd.DataFrame({
    "student_id": np.arange(1, n + 1),
    "city": rng.choice(["Kraków", "Poznań", "Warszawa", "Wrocław"], n),
    "school": rng.choice(schools_list, n),
    "grade": rng.integers(1, 5, n),
    **{f"task_{i}": rng.integers(0, 101, n).astype(float) for i in range(1, 6)},
    "python_years": rng.integers(0, 6, n),
    "submitted_at": (pd.Timestamp("2025-12-01") + pd.to_timedelta(rng.integers(0, 56 * 24 * 60, n), unit="min")).astype(str),
})
for col in [f"task_{i}" for i in range(1, 6)]:
    df.loc[rng.random(n) < 0.05, col] = np.nan                  # some missing scores
df = pd.concat([df, df.sample(5, random_state=1)], ignore_index=True)  # 5 duplicated rows

schools = pd.DataFrame({
    "school": schools_list,
    "type": ["liceum", "liceum", "liceum", "technikum", "technikum"],
})
print(df.shape, schools.shape)
```
